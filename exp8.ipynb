```python
df['target']=df['exang']
print("\nCreated 'target' from 'exang'. Values:",sorted(df['target'].unique()))
```

```python
if df['age'].nunique()>10:
    df['age']=pd.cut(df['age'],bins=5,labels=False)
if df['chol'].nunique()>10:
        df['chol']=pd.cut(df['chol'],bins=5,labels=False)
```

```python
from pgmpy.models import DiscreteBayesianNetwork
from pgmpy.parameter_estimator import DiscreteMLE
```

```python
models=DiscreteBayesianNetwork([
    ('age','target'),
    ('sex','target'),
    ('cp','target'),
    ('exang','target'),
    ('target','restecg'),
    ('target','chol')
])
```

```python
models.fit(df,estimator=DiscreteMLE())
print("\nBayesian Network Created Successfully")
```

```python
print("\nCPD of AGE\n")
print(models.get_cpds('age'))
```

```python
print("\nCPD of SEX\n")
print(models.get_cpds('age'))
```

```python
print("\nCPD of EXANG\n")
print(models.get_cpds('exang'))
```

```python
print("\nCPD of TARGET\n")
print(models.get_cpds('target'))
```

```python
print("\nCPD of CHOLESTEROL\n")
print(models.get_cpds('chol'))
```

```python
print("\nModel Check\n")
print(models.check_model())
```

```python
print("\nNodes")
print(models.nodes())
print("\nEdges")
print(models.edges())
```
